LOSO ON LR

```python
from sklearn.model_selection import LeaveOneGroupOut
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
import numpy as np
import pandas as pd
import os
import kagglehub

# Re-establishing data_path and reloading data with all transformations
# This path is based on the output from cell EtppLk_wCe2z and 8CAqKrCFCe3S
uciml_human_activity_recognition_with_smartphones_path = kagglehub.dataset_download('uciml/human-activity-recognition-with-smartphones')
data_path = '/root/.cache/kagglehub/datasets/uciml/human-activity-recognition-with-smartphones/versions/2'

# Re-load the original data
data = pd.read_csv(os.path.join(data_path, "train.csv"))

# Re-apply transformations to data to match the state before X and y were created
# 1. Drop duplicated columns (logic from cell PrZy-kbMCe3d and GQfb0XMuCe3f)
duplicated_columns = data.columns[data.T.duplicated()].tolist()
data = data.drop(duplicated_columns, axis=1)

# 2. Convert 'Activity' to categorical codes (logic from cell 9iFbfiHpCe3k)
data['Activity'] = data['Activity'].astype('category').cat.codes

# 3. Feature engineering (logic from cell cVPt5H4FCe4A)
data["acc_mag"] = np.sqrt(
    data["tBodyAcc-mean()-X"]**2 +
    data["tBodyAcc-mean()-Y"]**2 +
    data["tBodyAcc-mean()-Z"]**2
)
data["gyro_mag"] = np.sqrt(
    data["tBodyGyro-mean()-X"]**2 +
    data["tBodyGyro-mean()-Y"]**2 +
    data["tBodyGyro-mean()-Z"]**2
)
data["movement_score"] = data["acc_mag"] + data["gyro_mag"]

# Now proceed with the original LOSO code
logo = LeaveOneGroupOut()

lr_pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('model', LogisticRegression(max_iter=1000, solver='liblinear'))
])

lr_scores = []

groups = data['subject'] # Define 'groups' using the 'subject' column

for i, (train_idx, test_idx) in enumerate(logo.split(X, y, groups)):

    print(f"Fold {i+1}")

    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

    lr_pipeline.fit(X_train, y_train)

    y_pred = lr_pipeline.predict(X_test)

    lr_scores.append(accuracy_score(y_test, y_pred))

print("LR LOSO Accuracy:", np.mean(lr_scores))
```

```python
from sklearn.model_selection import LeaveOneGroupOut
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
import numpy as np


svm_pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('model', SVC(kernel='linear'))
])

svm_scores = []

groups = data['subject'] # Define 'groups' using the 'subject' column

for train_idx, test_idx in logo.split(X, y, groups):

    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

    svm_pipeline.fit(X_train, y_train)
    y_pred = svm_pipeline.predict(X_test)

    svm_scores.append(accuracy_score(y_test, y_pred))

print("SVM LOSO Accuracy:", np.mean(svm_scores))
```
